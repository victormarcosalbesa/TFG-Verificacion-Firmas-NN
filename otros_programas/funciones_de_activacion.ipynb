# Funciones de activación

El siguiente código grafica las diferentes funciones de activación mencionadas en el trabajo.

```python
"""
Graficado comparativo de funciones de activación
"""

import numpy as np
import matplotlib.pyplot as plt

# Generación de datos
z_full = np.linspace(-2.5, 2.5, 1000)

# Cálculo de las funciones continuas
sigmoid = 1 / (1 + np.exp(-z_full))
tanh = np.tanh(z_full)
relu = np.maximum(0, z_full)

# Configuración de la figura
plt.figure(figsize=(12, 7))

# Graficado de las funciones continuas
plt.plot(z_full, sigmoid, label='Sigmoide', color='blue', linewidth=3, alpha=0.8)
plt.plot(z_full, tanh, label='Tanh', color='green', linewidth=3, alpha=0.8)
plt.plot(z_full, relu, label='ReLU', color='purple', linewidth=3, alpha=0.8)

# Graficado de la función Heaviside

# Tramo izquierdo
plt.plot([-2.5, 0], [0, 0], color='red', linewidth=3, label='Heaviside')

# Tramo derecho
plt.plot([0, 2.5], [1, 1], color='red', linewidth=3)

plt.plot(0, 0, marker='o', color='red', markersize=9, markerfacecolor='red', markeredgecolor='red')

plt.plot(0, 1, marker='o', color='red', markersize=9, markerfacecolor='white', markeredgecolor='red')

# Configuración de los límites de los ejes
plt.xlim([-2.5, 2.5])
plt.ylim([-1.2, 2])

plt.xticks(np.arange(-2.5, 2.6, 0.5))
plt.yticks(np.arange(-1.2, 2.1, 0.4))
plt.xlabel('z', fontsize=16)
plt.ylabel('f(z)', fontsize=16)

plt.axhline(0, color='black', linewidth=1.2, alpha=0.9)
plt.axvline(0, color='black', linewidth=1.2, alpha=0.9)

plt.grid(True, linestyle='--', color='gray', alpha=0.5)
plt.legend(loc='upper left', fontsize=16, frameon=True, shadow=True)

plt.tight_layout()
plt.show()
```
